This shows the outliers in terms of the stress per emergency department. For some questions, it is important to not include these instances, but for others, it is imperative to include them.

```python
#Try to find outliers in the dataset such as the following,
# show examples, and show how you are cleaning them.

#Find Outliers of the emergency department stress
hospitalStress = ((data["Tot_ED_NmbVsts"] / data["EDStations"]) - (data["Tot_ED_NmbVsts"] / data["EDStations"]).mean()) / (data["Tot_ED_NmbVsts"] / data["EDStations"]).std()
print(hospitalStress)

#sns.histplot(hospitalStress)

outliers = hospitalStress[np.abs(hospitalStress) > 3]
print(outliers)
```
